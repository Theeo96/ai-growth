Ridge·Lasso 규제은 결과값만 얻는 것보다 입력 형태와 가정을 확인하고, 처리 전후를 비교하는 습관이 중요합니다.

### 구현 답안

```python
def solve(X_train, y_train, X_test, y_test):
    from sklearn.linear_model import Ridge,Lasso
    out={}
    for name,model in {'ridge':Ridge(alpha=1.0),'lasso':Lasso(alpha=0.1)}.items():
        model.fit(X_train,y_train); out[name]=mean_absolute_error(y_test,model.predict(X_test))
    return out
```
