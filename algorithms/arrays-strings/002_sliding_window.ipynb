### 구현 답안

```python
def solve(values, k):
    window=sum(values[:k]); best=window
    for i in range(k,len(values)):
        window+=values[i]-values[i-k]; best=max(best,window)
    return best
```

### 연관 알고리즘 문제 답안

```python
def solve_secondary(text):
    from collections import Counter
    counts=Counter(text)
    return min(counts, key=lambda c:(-counts[c],c))
```

```python

```
