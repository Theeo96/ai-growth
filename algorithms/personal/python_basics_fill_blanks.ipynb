# 초급 Python 빈칸 연습: 짝수의 합 구하기

리스트, 변수, `for` 반복문, `if` 조건문을 연습합니다. 현재 `.py`에 작성한 풀이를 옮겼습니다. 원래 빈칸은 코드 주석으로 남기고 바로 근처에 정답과 설명을 표시합니다.

Python 3 커널을 선택하고 위에서부터 실행하거나 **Run All(모두 실행)**을 누르세요. 별도 라이브러리는 필요 없습니다. 예상 출력은 `짝수의 합: 6`입니다.

## 1. 입력과 합계 초기화

`numbers`는 확인할 숫자 목록입니다. 합계는 아직 더한 숫자가 없으므로 0부터 시작합니다.

```python
numbers = [1, 2, 3, 4, 5]

# 빈칸 1: total = ___
# 정답: 0 — 합계를 0으로 시작합니다.
total = 0
```

## 2. 짝수만 합계에 더하기

리스트에서 숫자를 하나씩 꺼냅니다. `%`는 나머지를 구하며, 2로 나눈 나머지가 0이면 짝수입니다. 들여쓰기 안의 덧셈은 짝수일 때만 실행됩니다.

이 셀을 다시 실행할 때는 위의 초기화 셀도 먼저 실행하세요. 그렇지 않으면 이전 합계에 다시 더해집니다.

```python
# 빈칸 2: for number in ___:
# 정답: numbers — 반복할 리스트입니다.
for number in numbers:
    # 빈칸 3: if number % 2 == ___:
    # 정답: 0 — 2로 나눈 나머지가 0인지 확인합니다.
    if number % 2 == 0:
        # 빈칸 4: total = total + ___
        # 정답: number — 현재 짝수를 합계에 더합니다.
        total = total + number
```

## 3. 결과 확인

짝수 2와 4가 더해져 최종 합계는 6입니다.

```python
print("짝수의 합:", total)
```
